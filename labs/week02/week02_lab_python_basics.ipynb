## Part 4 — Floating point

Floats are binary approximations of decimal numbers.

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

Never compare floats with `==`. Compare the magnitude of the
difference against a tolerance:

$$|a - b| < \epsilon$$

You'll use exactly this test in Week 4 to decide when an iterative solver has
converged.

````python
tolerance = 1e-9
print(abs((0.1 + 0.2) - 0.3) < tolerance)
````

## Part 5 — Lists, indexing, slicing

A year of monthly water-level measurements, January first.

````python
monthly_head = [
    415.2, 414.8, 414.1, 413.2, 412.0, 410.6,
    409.1, 411.7, 413.9, 414.6, 415.0, 415.3,
]

print(f"number of measurements: {len(monthly_head)}")
print(f"January (index 0):      {monthly_head[0]}")
print(f"December (index -1):    {monthly_head[-1]}")
print(f"lowest level:           {min(monthly_head)}")
````

Slices are **half-open**: `[start:stop]` includes `start` and excludes
`stop`. So `stop - start` is always the number of elements you get.

````python
print(f"first quarter:  {monthly_head[0:3]}")
print(f"same thing:     {monthly_head[:3]}")
print(f"from July on:   {monthly_head[6:]}")
print(f"last two:       {monthly_head[-2:]}")
````

## Part 6 — Tuples and dictionaries

maps names to values.

````python
grid_shape = (3, 40, 60)   # layers, rows, columns — a MODFLOW grid
n_layers, n_rows, n_cols = grid_shape   # unpacking
print(f"{n_layers} layers, {n_rows} rows, {n_cols} columns")
print(f"total cells: {n_layers * n_rows * n_cols}")

# grid_shape[0] = 5 would raise TypeError — tuples are immutable
````
